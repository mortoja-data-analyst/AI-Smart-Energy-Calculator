```python
# ===================================================
# PROJECT 10: AI SMART ENERGY & ELECTRICITY CALCULATOR
# AUTHOR: MORTOJA SARDAR (DATA ANALYST)
# ===================================================

!pip install gTTS -q
import time, IPython.display as ipd
from gtts import gTTS

print("=" * 60)
print("     ⚡ ENTERPRISE AI SMART ENERGY CALCULATOR ⚡     ")
print("=" * 60 + "\n")

total_units_consumed = 420.0 # পুড়ে যাওয়া বিদ্যুতের ইউনিট
cost_per_unit_usd = 0.15     # প্রতি ইউনিটের দাম ০.১৫ ডলার

print(f"📥 [METER DATA] Total Consumption Ingested: {total_units_consumed} Units")
print("📊 [SYSTEM STATUS] Processing mathematical tariff and utility rates...")
time.sleep(2)

total_bill_amount = total_units_consumed * cost_per_unit_usd

if total_units_consumed > 300.0:
    consumption_rating = "HIGH ENERGY CONSUMPTION WARNING ⚠️"
    voice_msg = f"Alert Mortoja! Electricity consumption is critically high at {int(total_units_consumed)} units. Total generated bill is {int(total_bill_amount)} dollars. Please activate power-saving mode."
else:
    consumption_rating = "OPTIMAL ENERGY CONSUMPTION COMPLIANT ✅"
    voice_msg = f"Success Mortoja. Energy consumption is optimal. Total bill is {int(total_bill_amount)} dollars."

print("-" * 60)
print(f"💰 POWER CONSUMED :  {total_units_consumed} Units")
print(f"🌍 TARIFF RATE    :  $ {cost_per_unit_usd} per Unit")
print("-" * 60)
print(f"🏆 TOTAL UTILITY  :  $ {total_bill_amount:,.2f} USD")
print(f"📈 USAGE STATUS   :  {consumption_rating}")
print("-" * 60 + "\n")

tts = gTTS(voice_msg, lang='en')
tts.save("energy_bill_report.mp3")
ipd.Audio("energy_bill_report.mp3")
```
